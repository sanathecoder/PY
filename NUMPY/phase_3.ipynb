### Data Operation with Bussiness Example

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure : [resturant_id, 2021, 2022, 2023, 2024]

sale_data =  np.array([
    [1, 150000, 180000, 220000, 250000],  #Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  #Beijing Bite
    [3, 200000, 230000, 260000, 270000],  #Pizza Hub
    [4, 180000, 210000, 240000, 270000],  #Burgure Point
    [5, 160000, 185000, 205000, 230000]   #Chai Point
])

print("=== Zomato sales Analysis ===")
print(" Sale Data Shap: ", sale_data.shape )
print(" Sample data for 1st  3 resturant: \n", sale_data[:3])
```

```python
# Total Sale Per Year

print(np.sum(sale_data, axis=0))

year_sale = np.sum(sale_data[:,1:], axis=0)
print(year_sale)

# Minimun Sale per Resturant 
min_sale = np.min(sale_data[0:,1:], axis=1)
print(min_sale)

# Max Sale per Resturant
max_sale = np.max(sale_data[0:,1:], axis=1)
print(max_sale)

#Average Sale per Resturant
avg_sale = np.mean(sale_data[:,1:], axis=1)
print(avg_sale)

```

```python
cum_sale = np.cumsum(sale_data[:,1:], axis=1)
print(cum_sale)

plt.figure(figsize=(8,6))
plt.plot(np.mean(cum_sale, axis=0))
plt.title("Average Cumulative Sale  across all resturant")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])
print("Vector Addition: ", vector1 + vector2)
print("Vector Multiplication: ", vector1 * vector2)
print("Dot Product: ", np.dot(vector1,vector2))

angle = np.arccos(np.dot(vector1, vector2)/(np.linalg.norm(vector1) * np.linalg.norm(vector2) ))
print(angle)


```
