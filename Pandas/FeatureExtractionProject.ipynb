```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('anime.csv')
```

```python
def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data += i
        if i == "(":
            check = True
    return i
```

```python
df['Episode'] =  df['Title'].apply(extract_episodes)
```

```python
df['Episode'].head()
```

```python
df['Episode'] =  df['Episode'].str.replace(" eps","")
```

```python
df['Episode'] = df['Episode'].astype(int)
```

```python
df.head()
```

```python
df.loc[2]['Title']
```

```python
def extract_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i+1,i+20):
                data += txt[j]
            return data
        
```

```python
df['Total Time'] =  df['Title'].apply(extract_time)
```

```python

def calculate_months(x):
    start, end = x.split(' - ')
    
    start = pd.to_datetime(start, format='%b %Y')
    end = pd.to_datetime(end, format='%b %Y')
    
    return (end.year - start.year) * 12 + (end.month - start.month)

df['Months'] = df['Total Time'].apply(calculate_months)
```
