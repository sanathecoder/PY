### Find Missing Data

```python
data = {
    'A':[1,2,np.nan , 4,5],
    'B': [np.nan, 2,3,4,5],
    'C': [1,2,3,np.nan,5],
    'D': [1,np.nan, np.nan,np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
df.dropna()
```

```python
df.dropna(thresh=3)
```

### Fill Missing Values 

```python
df.fillna(0)
```

```python
values = {'A':10,'B':20,'C':30,'D':40}
df.fillna(values)
```

```python
df.fillna(df.mean())
```

### Merging and joining DataFrame

```python
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    'name': ['Sana','Amna','Nabiha','Hafsa','Rabia'],
    'department': ['HR','IT','IT','HR','Finance']
})


salaries  = pd.DataFrame({
    'employee_id': [1,2,3,6,7],
    'salary': [80000,90000,60000,40000,50000],
    'bonous' : [2000,4000,6000,9000,3000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries, on='employee_id')
```
