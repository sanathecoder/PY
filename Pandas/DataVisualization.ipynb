```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(1,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Graph")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
```

### Subplot

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(y,y)
plt.subplot(2,2,4)
plt.plot(y,x)
```

```python
fig = plt.figure(figsize=(10,8) )
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```
