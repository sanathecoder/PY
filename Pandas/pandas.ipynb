```python
# Data Cleaning
df.dropna(subset = ["city", "age"])
```

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
df['joining_date'] = pd.to_datetime(df['joining_date'])
```

```python
df["annual_salary"] = df["salary"] * 12
```

```python
df.head()
```

```python
def catagorize(num):
    if num > 5:
        return "Experienced"
    elif num <= 5:
        return " Learner"
```

```python
df["experience_status"]  = df["experience"].apply(catagorize)
df.head()
```

```python
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")

df["salary_level"] = df["salary"].apply(
    lambda salary: "High" if salary > 70000 else "Normal"
)
```

```python
department_code = {"Engineering": "ENG","Marketing": "MKT", "Finance": "FIN","Data Science":"DS"}
df["department_code"] = df["department"].map(department_code)
df.head()
```

```python
df["age"] = df["age"].fillna(df["age"].mean()).astype(int)
```
